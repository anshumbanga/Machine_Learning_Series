```python
# importing major libraries

import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('placement.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
# independent columns

x = df[['cgpa']]
x = df.iloc[:,0:-1]
x.ndim
```

```python
# dependent columns
# series, not data frames

y = df['package']
y = df.iloc[:,-1]
y.ndim
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
y_train.head()
```

```python
# patterns
# matplotlib
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('cgpa')
plt.ylabel('package')
plt.title('cgpa vs package')
plt.show()
```

```python
# linearregression

from sklearn.linear_model import LinearRegression
```

```python
# object of this class

lr = LinearRegression()
```

```python
lr
```

```python
# train
# fit function

lr.fit(x_train,y_train)
```

```python
m = lr.coef_
c = lr.intercept_
```

```python
print('Slope\t\t:',m)
print('Intercept\t:',c)
```

```python
# testing
# predict

y_pred = lr.predict(x_test)

y_pred
```

```python
x_test.iloc[0]
```

```python
input_x = 6.63

m * input_x + c
```

```python
# answer key --> y_test
# asnwer sheet --> y_pred
```

```python
results = pd.DataFrame({'Actual':y_test,'Predicted':y_pred,'error':y_test-y_pred})

results
```

```python
# patterns
# matplotlib
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
plt.scatter(df['cgpa'],df['package'],label='Training Data')
plt.scatter(x_test,y_test,label='Testing Data')
plt.plot(x_test,y_pred,color='red',label='Regression Line')
plt.legend()
plt.xlabel('cgpa')
plt.ylabel('package')
plt.title('cgpa vs package')
plt.show()
```

```python
plt.figure(figsize=(8,4))
# plt.scatter(df['cgpa'],df['package'],label='Training Data')
plt.scatter(x_test,y_test,label='Actual Point')
plt.plot(x_test,y_pred,color='red',label='Predicted')
plt.legend()
plt.xlabel('cgpa')
plt.ylabel('package')
plt.title('cgpa vs package')
plt.show()
```

```python
# metrics module of sklearn

from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
```

```python
print('Mean Absolute Error\t\t:',mean_absolute_error(y_test,y_pred),'lakhs')
print('Mean Squared Error\t\t:',mean_squared_error(y_test,y_pred),'lakhs^2')
print('Root Mean Squared Error\t\t:',np.sqrt(mean_squared_error(y_test,y_pred)),'lakhs')
```

```python
print('Coeff of Determination\t\t:',r2_score(y_test,y_pred)*100,'%')
```
